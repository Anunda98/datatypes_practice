```python
puppy_data = [
    ['Name', 'Breed', 'Good Doggy?', 'Length [in]', 'Weight Status', 'Body Temp [F]'],
    ['Rocky', 'labrador_retriever', 'Good', 25, 'Overweight', 101.0],
    ['Molly', 'French_bulldog', 'Bad', 12, 'Normal', 103.0],
    ['Toby', 'beagle', 'Good', 15, 'Underweight', 100.5],
    ['Sadie', 'miniature_schnauzer', 'Good', 14, 'Normal', 102.5],
    ['Oscar', 'pomeranian ', 'Bad', 10, 'Normal', 101.5],
    ['Ruby', 'german_shepherd', 'Good', 26, 'Overweight', 104.0],
    ['Finn', 'Chihuahua', 'Bad', 9, 'Underweight', 99.9],
    ['Coco', 'pug', 'Good', 11, 'Normal', 102.0],
]
```

```python
#Clean the Breed names

#first create a function to perform all the above
def clean_string(text):
    text = text.strip()                 #Remove leading/trailing spaces
    text = text.replace("_"," ")        #Replace underscores _ with spaces
    text = text.title()                 #Capitalize the first letter of each word
    return text
```

```python
#Apply it to YOUR dataset

for i, row in enumerate(puppy_data):
    if i > 0:  # skip header
        row[1] = clean_string(row[1])
puppy_data
```

```python
#Convert Length from inches to centimeters
#lenght -index3

for i, row in enumerate(puppy_data):
    if i > 0:
        row[3] = row[3] * 2.54

#checking if length is converted
for row in puppy_data:
    print(row[3])
```

```python
#changing the header
puppy_data[0][3] = 'Length [cm]'
puppy_data[0][3] 
```

```python
#Convert Good/Bad doggy column to Boolean
for i, row in enumerate(puppy_data):
    if i > 0:
        if row[2] == "Good":
            row[2] = True
        else:
            row[2] = False
            
#running checks
for row in puppy_data:
    print(row[2])
```

```python
#Convert Weight Status to integers
    #Mapping: 'Underweight': 0, 'Normal': 1, 'Overweight': 2, 'Obese': 3
weight_status = {'Underweight': 0, 'Normal': 1, 'Overweight': 2, 'Obese': 3}
for i, row in enumerate(puppy_data):
    if i > 0:
        row[4] = weight_status[row[4]]

#running checks
for row in puppy_data:
    print(row[4])
```

```python
#Create a new column Has_Fever
puppy_data[0].append('Has Fever')
```

```python
#True if Body Temp > 102 F, else False

for i, row in enumerate(puppy_data):
    if i > 0:
        row.append(row[5] > 102)   # yields True if >102, False otherwise
```
